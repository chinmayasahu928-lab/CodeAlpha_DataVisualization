```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


from google.colab import files

print("Please upload your Task 1 CSV file:")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("\nFile uploaded:", file_name)



df = pd.read_csv(file_name)

print("\n" + "=" * 60)
print("TASK 3 - DATA VISUALIZATION")
print("=" * 60)

print("Total Books:", len(df))



rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Number"] = df["Rating"].map(rating_map)



rating_order = [
    "One",
    "Two",
    "Three",
    "Four",
    "Five"
]

rating_counts = df["Rating"].value_counts().reindex(
    rating_order
)

plt.figure(figsize=(8, 5))

plt.bar(
    rating_counts.index,
    rating_counts.values
)

plt.title("Number of Books by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Books")

plt.show()


plt.figure(figsize=(8, 5))

plt.hist(
    df["Price"],
    bins=20
)

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.show()



top_expensive = df.sort_values(
    "Price",
    ascending=False
).head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_expensive["Book Title"],
    top_expensive["Price"]
)

plt.title("Top 10 Most Expensive Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")

plt.gca().invert_yaxis()

plt.show()



top_cheap = df.sort_values(
    "Price",
    ascending=True
).head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_cheap["Book Title"],
    top_cheap["Price"]
)

plt.title("Top 10 Cheapest Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")

plt.show()



average_price = df.groupby(
    "Rating"
)["Price"].mean().reindex(
    rating_order
)

plt.figure(figsize=(8, 5))

plt.bar(
    average_price.index,
    average_price.values
)

plt.title("Average Book Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Price (£)")

plt.show()



plt.figure(figsize=(8, 5))

plt.scatter(
    df["Rating_Number"],
    df["Price"]
)

plt.title("Price vs Rating")
plt.xlabel("Rating")
plt.ylabel("Price (£)")

plt.show()



plt.figure(figsize=(6, 5))

plt.boxplot(
    df["Price"].dropna()
)

plt.title("Book Price Distribution - Box Plot")
plt.ylabel("Price (£)")

plt.show()



print("\n" + "=" * 60)
print("VISUALIZATION SUMMARY")
print("=" * 60)

print("Total Books:", len(df))

print(
    "Average Price: £",
    round(df["Price"].mean(), 2)
)

print(
    "Highest Price: £",
    round(df["Price"].max(), 2)
)

print(
    "Lowest Price: £",
    round(df["Price"].min(), 2)
)

print(
    "Average Rating:",
    round(df["Rating_Number"].mean(), 2)
)

print(
    "Most Common Rating:",
    df["Rating"].value_counts().idxmax()
)

print("\nTASK 3 DATA VISUALIZATION COMPLETED! ✅")
```
